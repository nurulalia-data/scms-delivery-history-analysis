## Missing Value Handling

### Shipment Mode Missing Values

```python
df_delivery.groupby("Country")["Shipment Mode"].apply(
    lambda x: x.isna().sum()
).sort_values(ascending=False).head(12)
```

```python
df_delivery["Shipment Mode"].value_counts()
```

### Shipment Mode Missing Value Handling

```python
df_delivery["Shipment Mode"] = df_delivery["Shipment Mode"].fillna("Unknown")
```

### Dosage Missing Value Handling

```python
df_delivery.groupby("Product Group")["Dosage"].apply(lambda x: x.isna().sum()).sort_values(ascending=False)
```

```python
# Replace missing dosage values with "Not Provided"
```

```python
df_delivery["Dosage"] = df_delivery["Dosage"].fillna("Not Provided")
```

```python
df_delivery["Dosage"].value_counts().head(10)
```

### Delivery Delay and Status Transformation

```python
df_delivery["Delivery Delay Days"] = (
    df_delivery["Delivered to Client Date"]
    - df_delivery["Scheduled Delivery Date"]
).dt.days
```

```python
# Classify shipments based on delivery delay days
```

```python
df_delivery["Delivery Status"] = df_delivery["Delivery Delay Days"].apply(
    lambda x: "Early" if x < 0
    else "On Time" if x == 0
    else "Delayed"
)
```

```python
df_delivery["Delivery Status"].value_counts()
```

### Delivery Delay Rate by Country

```python
delayed_rate = (
    df_delivery.groupby("Country")["Delivery Status"]
    .apply(lambda x: (x == "Delayed").mean() * 100)
    .sort_values(ascending=False)
)

delayed_rate.head(10)
```

#### Delivery Delay Rate for Countries with Sufficient Sample Size

Countries with at least 50 shipments are included to provide a more reliable comparison of delivery delay rates.

```python
delayed_rate_50 = (
    df_delivery.groupby("Country")
    .filter(lambda x: len(x) >= 50)
    .groupby("Country")["Delivery Status"]
    .apply(lambda x: (x == "Delayed").mean() * 100)
    .sort_values(ascending=False)
)

delayed_rate_50
```

### Country Delivery Performance Summary

```python
country_delivery = (
    df_delivery.groupby("Country")
    .agg(
        Shipments=("ID", "count"),
        Delayed=("Delivery Status", lambda x: (x == "Delayed").sum())
    )
)

country_delivery["Delayed Rate (%)"] = (
    country_delivery["Delayed"] / country_delivery["Shipments"] * 100
)

country_delivery.sort_values(
    "Delayed Rate (%)",
    ascending=False
).head(10)
```

### Top Countries by Delivery Delay Rate

```python
country_delivery[country_delivery["Shipments"] >= 50].sort_values(
    "Delayed Rate (%)",
    ascending=False
).head(10)
```

### Delivery Delay Rate by Shipment Mode

Assess delivery delay rates across different shipment modes.

```python
shipment_mode_delivery = (
    df_delivery.groupby("Shipment Mode")["Delivery Status"]
    .value_counts(normalize=True)
    .mul(100)
    .unstack(fill_value=0)
)

shipment_mode_delivery
```
